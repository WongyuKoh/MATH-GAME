```python
def findtrytime(text):

	#text=data_parsing['user_answers']
	text=text.replace(',,',',')
	text=text.split(',')

	trytime = -1
	if len(text)==0:
		if len(text)==0:
			trytime=5
		elif text=="ac":
			trytime=1
		else:
			trytime=5
	else:
		if text[-1]=='wa' or text[-1]=='w':
			trytime=5
		elif text[-1]=='ac' or text[-1]=='c':
			trytime=len(text)
		else:
			if text[-1]=='':
				trytime=5
			else:
				trytime=len(text)
    

	if trytime>4:
		trytime=5

	return trytime
```

```python
def number_parsing(pb):
	if pb[1]=='(':
		return int(pb[2:][:-2])
	else:
		return int(pb[1:][:-1])
```

```python
def text_parsing(pb):
	if pb[1]=='(':
		return pb[2:][:-2]
	else:
		return pb[1:][:-1]
```
